```python
print("Mean:", df['Profit'].mean())
print("Sum:", df['Profit'].sum())
print("90% Quantile:",df['Profit'].quantile(0.90))
```

```python
print(df['Category'].value_counts())
```

```python
df[df['Customer ID']=='RB-19465']['Customer Name'].unique()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True, format='mixed')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], dayfirst=True, format='mixed')
print("Order Dates:", df['Order Date'].min(), "to", df['Order Date'].max())
print("Ship Dates:", df['Ship Date'].min(), "to", df['Ship Date'].max())
```

```python
from numpy import isin
south_dates = df[df['Region'] == 'South']['Order Date']
west_dates = df[df['Region'] == 'West']['Order Date']
matching_days = set(south_dates).intersection(set(west_dates))
order_same_day = df[df['Order Date'].isin(matching_days)]
```

```python
df.groupby('Region')['Profit'].mean()
```
